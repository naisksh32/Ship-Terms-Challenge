### 구분자를 ||로 하면 오류가 생겨서 |로 구분자를 바꾸는 코드

```python
import pandas as pd
import os
```

```python
def change_delimiter(input_file, output_file):
    """
    파일의 구분자를 '||'에서 '|'로 변경하여 새 파일에 저장하는 함수.
    
    Args:
        input_file (str): 구분자를 변경할 원본 파일 경로.
        output_file (str): 수정된 내용을 저장할 새 파일 경로.
    """
    try:
        with open(input_file, 'r', encoding='utf-8') as f_in:
            with open(output_file, 'w', encoding='utf-8') as f_out:
                for line in f_in:
                    # 각 줄의 ||를 |로 변경
                    modified_line = line.replace('||', '|')
                    f_out.write(modified_line)
        print(f"✅ 파일이 성공적으로 '{output_file}'에 저장되었습니다.")
        
    except FileNotFoundError:
        print(f"❌ 오류: 원본 파일 '{input_file}'을 찾을 수 없습니다.")
    except Exception as e:
        print(f"❌ 오류가 발생했습니다: {e}")
```

```python
# 원본 파일 경로
input_path = '../../공공/Data/en_ko_pairs.txt'

# 수정된 내용을 저장할 새 파일 경로
output_path = '../../공공/Data/en_ko_pairs_modified.txt'

# 함수 실행
change_delimiter(input_path, output_path)
```
